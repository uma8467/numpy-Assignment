# C2. Z-score standardization:

#### C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy .


#### Explanation:

The mean is subtracted from each value, and the result is divided by the standard deviation. This example uses the population standard deviation (ddof=0), which is NumPy's default.

```python
def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std
x = np.array([10, 20, 30, 40, 50])
result = standardize(x)
print(result)
```

## D2. Image Batch Understanding

##### 1. Explain, with correct NumPy shapes, how a single RGB image differs from a batch of RGB images.

A single RGB image with a height and width of 64 × 64 has three color channels: Red, Green, and Blue. Therefore, its NumPy shape is `(64, 64, 3)`, where the three dimensions represent height, width, and color channels.

When multiple RGB images are stored together as a batch, an additional dimension is added at the beginning to represent the number of images. Therefore, a batch of 16 RGB images has the shape `(16, 64, 64, 3)`.

Here, `16` represents the number of images, `64 × 64` represents the height and width of each image, and `3` represents the RGB channels.

##### 2. Write code to create a dummy batch of 16 images of size 64 × 64 using random pixel values.

The following code creates a batch of 16 RGB images with dimensions 64 × 64. Each pixel is represented using three channels: Red, Green, and Blue.

```python
#Step 1: Create a dummy batch:
import numpy as np
rng = np.random.default_rng(42)
images = rng.integers(
    0, 256,
    size=(16, 64, 64, 3),
    dtype=np.uint8
)
print("Batch shape:", images.shape)

#Step 3: Calculate the average image:
average_image = images.mean(axis=0)
print("Average image shape:", average_image.shape)
```

## D3. Debug the Model Code

### Bug 1: Weight vector shape

`X` has shape `(3, 2)`, which means there are 3 samples and 2 features. Therefore, the weight vector must contain 2 values and have shape `(2,)`.

The original `w` has shape `(3,)`. This makes `X @ w` invalid because matrix multiplication would require `(3, 2) @ (2,)`, but the code attempts `(3, 2) @ (3,)`.

### Bug 2: Incorrect MSE calculation

The original expression is:

`np.mean(y_true - y_pred ** 2)`

Because exponentiation has higher precedence than subtraction, Python evaluates this as:

`np.mean(y_true - (y_pred ** 2))`

For Mean Squared Error, the difference must be calculated first and then squared. Therefore, the correct expression is:

`np.mean((y_true - y_pred) ** 2)`

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])
w = np.array([0.5, 1.0])
b = 2.0
y_pred = X @ w + b
y_true = np.array([10, 20, 30])
mse = np.mean((y_true - y_pred) ** 2)
print("Predictions:", y_pred)
print("MSE:", mse)
```

## Explanation:
The first bug is the incorrect shape of the weight vector. Since `X` has 2 features, `w` must contain 2 weights. The corrected weight vector therefore has shape `(2,)`.

The second bug is the MSE formula. Due to operator precedence, `y_pred ** 2` is calculated before the subtraction in the original code. Adding parentheses ensures that the prediction error is calculated first and then squared.

After correcting both bugs, the model produces one prediction for each of the 3 samples, and the mean of the squared errors gives the final MSE.

## E1. Per-Column Min–Max Normalization

Min-max normalization scales each feature to a range between 0 and 1. Since this question requires per-column normalization, the minimum and maximum values are calculated separately for each feature column using `axis=0`.

The normalization formula is:

`(X - column_min) / (column_max - column_min)`

NumPy broadcasting automatically applies each column's minimum and maximum to the corresponding column. Therefore, the entire operation is vectorized and does not require explicit loops over rows or columns.

```python
X = np.array([[10, 100],
              [20, 200],
              [30, 300]], dtype=float)
col_min = X.min(axis=0)
col_max = X.max(axis=0)
normalized = (X - col_min) / (col_max - col_min)
print(normalized)
```

### Explanation:
 The minimum and maximum are calculated separately for each column. Broadcasting applies each column's scaling parameters to its values.

## E2. Three-Level Score Classification:

The exam scores need to be classified into three categories based on their values. Scores below 40 are classified as "Fail", scores from 40 to 70 inclusive are classified as "Average", and scores above 70 are classified as "Excellent".

I use nested `np.where()` to perform all three classifications in a single vectorized expression. The first `np.where()` checks whether a score is below 40. If it is not, the second `np.where()` checks whether the score is less than or equal to 70. This avoids explicit loops and if/else statements.

```python
scores = np.array([25, 40, 55, 70, 71, 90])
labels = np.where(
    scores < 40,
    "Fail",
    np.where(
        scores <= 70,
        "Average",
        "Excellent"
    )
)
print(labels)
```

### Explanation:
The outer np.where() identifies scores below 40. For the remaining scores, the inner np.where() separates Average (up to and including 70) from Excellent (above 70).
